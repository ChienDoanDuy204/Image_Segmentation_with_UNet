```python
import torch
import torch.nn as nn
torch.manual_seed(42)
```

# Deconvolution - Transpose Convolution

```python
TransposeConv = nn.ConvTranspose2d(
    in_channels = 1,
    out_channels = 1,
    kernel_size = 2,
    bias = False,
)
```

```python
TransposeConv.weight.data = torch.ones(
    TransposeConv.weight.data.shape
)
```

```python
TransposeConv.weight
```

```python
input = torch.randint(
    5, (1,2,2), dtype = torch.float32
)
input
```

```python
ouput = TransposeConv(input)
ouput
```

# Padding in Transpose Convolution

```python
TransposeConv = nn.ConvTranspose2d(
    in_channels = 1,
    out_channels = 1,
    kernel_size = 2,
    bias = False,
    padding = 1
)
```

```python
TransposeConv.weight.data = torch.ones(
    TransposeConv.weight.data.shape
)
```

```python
output = TransposeConv(input)
output
```

# Stride in Transpose Convolution

```python
TransposeConv = nn.ConvTranspose2d(
    in_channels = 1,
    out_channels = 1,
    kernel_size = 2,
    bias = False,
    stride = 2,
)
```

```python
TransposeConv.weight.data = torch.ones(
    TransposeConv.weight.data.shape
)
```

```python
ouput = TransposeConv(input)
ouput
```

# Multi channels in Transpose Convolution

```python
Input = torch.randint(
    5, (2,2,2), dtype = torch.float32
)
Input
```

```python
TransposeConv = nn.ConvTranspose2d(
    in_channels = 2,
    out_channels = 2,
    kernel_size = 2,
    bias = False,
    stride = 2,
)
```

```python
TransposeConv.weight.data = torch.ones(
    TransposeConv.weight.data.shape
)
```

```python
TransposeConv.weight
```

```python
output = TransposeConv(Input)
output
```
